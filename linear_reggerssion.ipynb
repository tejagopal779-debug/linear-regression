```python
df=pd.read_csv(r"C:\Users\DELL\Downloads\melb_data.csv")
df
```

```python
df["Date"]=pd.to_datetime(df["Date"],format="mixed")
```

```python
df["year"]=df["Date"].dt.year
df

```

```python
df=df.drop(columns=["Date","Longtitude","Lattitude",])
df
```

```python
y =df["Price"]
x=df.drop(columns="Price")
```

```python
num=x.select_dtypes(include="number")
cate=x.select_dtypes(exclude="number")
```

```python
num1=Pipeline(steps=[("impute",SimpleImputer(strategy="mean")),("scaling",StandardScaler())])
cat1=Pipeline(steps=[("impute",SimpleImputer(strategy="most_frequert")),("scaling",OneHotEncoder())])
```

```python
pre=ColumnTransformer(transformers=[("num",num1,num),("cat",cat1,cate)],remainder="drop")
```

```python

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.10)



pre
```

```python
liner=Pipeline(steps=[("p",pre),("liner",LinearRegression())])
```
